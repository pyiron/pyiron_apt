<font style="font-family:roboto;color:#455e6c;font-size:18px;text-align:justify"> <b> Upload the data to NOMAD <br> </b> </font>
The NeXus/HDF5 and log files from pyiron are digital artifacts. They can be used for any type of further studies and archival purposes.<br>
Typical examples are uploads to research databases like [Zenodo](https://www.zenodo.org), [Coscine](https://about.coscine.de/), electronic lab notebooks,<br>
or research data management systems like [NOMAD](https://www.nomad-lab.eu). Different services offer different functionalities how<br>
these artifacts are then indexed and made searchable. In the case of NOMAD specific functionalities were implemented that offer an<br>
automated parsing of the content from NeXus/HDF5 files into the internal data structure and search system of NOMAD.

<font style="font-family:roboto;color:#455e6c;font-size:18px;text-align:justify"></font>
These functionalities are triggered by uploading the NeXus files [programmatically via making API calls to NOMAD](https://nomad-lab.eu/prod/v1/docs/howto/programmatic/publish_python.html).

Within the remaining time of the first funding period we plan to explore this route further exemplified for user data and using [NOMAD deployments](https://gitlab.mpcdf.mpg.de/nomad-lab/nomad-distro) of their own choice.<br>
These can be either the [official (central) NOMAD](https://nomad-lab.eu/prod/v1/gui/search/entries) where it is possible to make a data publication with DOI assignment, or local deployments.
